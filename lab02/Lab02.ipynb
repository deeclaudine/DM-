```python
import numpy as np
import pandas as pd

raw_data = {
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010, 1011, 1012, 1013, 1014, 1015],
    "channel": ["site", "app", "app", "marketplace", "site", "app", "site", "marketplace", "app", "site", "marketplace", "app", "site", "site", "marketplace"],
    "order_sum": [1200, 2500, 1800, 3200, 900, 1500, 2800, 45000, 2100, 1700, 3600, 1100, 2400, 1900, 3300],
    "delivery_time": [25, np.nan, 40, 55, 20, 30, np.nan, 70, 35, 45, 50, 18, 27, np.nan, 60],
    "rating": [5, 4, 5, 3, 5, 4, 2, 1, 5, 4, 3, 5, 4, 5, 2]
}

df = pd.DataFrame(raw_data)
print(df)
df.to_csv("online_store_orders.csv", index=False)
```

Для чего используется параметр index=False? Что произойдёт при сохранении таблицы без этого параметра?

Ответ: Параметр index=False запрещает pandas записывать индексы строк (0, 1, 2...) в CSV-файл как отдельный безымянный столбец. Если его не указать, при следующей загрузке файла появится лишний столбец Unnamed: 0, что нарушит структуру данных и усложнит дальнейший анализ.

```python
store_df = pd.read_csv("online_store_orders.csv")

print("Первые 5 строк:")
print(store_df.head())

print("\nПоследние 5 строк:")
print(store_df.tail())

print("\nРазмер таблицы:")
print(store_df.shape)

print("\nНазвания столбцов:")
print(store_df.columns.tolist())

print("\nИнформация о таблице:")
store_df.info()
```

Количество строк: 15

Столбцы с пропущенными значениями: delivery_time
Количество пропущенных значений: 3

```python
# 5.1 Выбор столбцов
print("Только сумма и время доставки:")
print(store_df[["order_sum", "delivery_time"]])

# 5.2 Выбор строк с индексами от 3 до 7 включительно
slice_df = store_df.iloc[3:8, [2, 3]]
print("\nСрез с индексами 3-7 (order_sum и delivery_time):")
print(slice_df)

# 5.3 Фильтрация
premium_orders = store_df[(store_df["order_sum"] > 2000) & (store_df["rating"] == 5)]
print("\nПремиум заказы (сумма > 2000 и рейтинг == 5):")
print(premium_orders)
```

Почему для объединения условий используется &, а не and?

Ответ: Оператор & является побитовым (векторизованным) и применяется к каждому элементу столбца (Series) по отдельности. Ключевое слово and в Python пытается оценить истинность всего объекта целиком, что для pandas Series вызывает ошибку ValueError, так как неясно, как интерпретировать истинность массива данных.

```python
# 6.1 Пропущенные значения
print("Пропущенные значения в каждом столбце:")
print(store_df.isna().sum())

# 6.2 Описательная статистика
print("\nОписательная статистика:")
print(store_df.describe())

# 6.3 Среднее и медиана времени доставки
mean_delivery = store_df["delivery_time"].mean()
median_delivery = store_df["delivery_time"].median()
print(f"\nСреднее время доставки: {mean_delivery:.2f}")
print(f"Медианное время доставки: {median_delivery:.2f}")

# 6.4 Распределение рейтингов
print("\nРаспределение рейтингов:")
print(store_df["rating"].value_counts())
```

1. Сколько заказов не имеют информации о времени доставки?
3 заказа.
2. Какой показатель имеет значение max, существенно отличающееся от остальных?
order_sum (максимум 45 000, что сильно выделяется на фоне остальных сумм ~1000–3000).
3. Почему среднее значение может отличаться от медианы?
 Среднее арифметическое чувствительно к выбросам (аномально высоким или низким значениям), которые "тянут" его в свою сторону. Медиана же показывает центральное значение упорядоченного ряда и устойчива к выбросам, поэтому лучше отражает "типичное" значение.
4. Какая оценка встречается наиболее часто? Оценка 5 (встречается 6 раз).

```python
# 7.1 Создание производного признака
store_df["is_high_value"] = store_df["order_sum"] > 3000
print("Проверка нового столбца:")
print(store_df[["order_id", "order_sum", "is_high_value"]])

# 7.2 Группировка по каналу продаж
channel_stats = store_df.groupby("channel")["order_sum"].mean().reset_index()
print("\nСредняя сумма заказа по каналам:")
print(channel_stats)

# 7.3 Сортировка результата
channel_stats = channel_stats.sort_values(by="order_sum", ascending=False)
print("\nОтсортированные каналы по убыванию средней суммы:")
print(channel_stats)
```

Средняя сумма заказа по каналам: marketplace (~13 775 руб.), site (~1 967 руб.), app (~1 800 руб.).

Количество заказов с is_high_value = True: 4 заказа.

Количество пропущенных значений: 3 (в столбце delivery_time).

Влияние выброса: Заказ на 45 000 руб. в канале "marketplace" искусственно завышает среднюю сумму по этому каналу и в целом по магазину.

Различие среднего и медианного времени доставки невелико (39.58 против 37.5), что говорит об относительно симметричном распределении времени доставки, несмотря на отдельные задержки.
